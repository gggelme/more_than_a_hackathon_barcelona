cuantos frames tiene el video: 

```python
VIDEO = games.require(["video"])[0]   # ruta al .mp4 táctico; si no existe, error claro
print(VIDEO)
```

```python
import cv2
cap = cv2.VideoCapture(VIDEO)
assert cap.isOpened(), "no se pudo abrir el video"

n_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)

print("frames:", n_frames)
print("fps:", fps)
print("duración:", n_frames / fps / 60, "min")
print("esperado:", games.source_frames(), "| coincide:", n_frames == games.source_frames())
```
